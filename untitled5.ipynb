```python
#Load libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
from tensorflow.keras.callbacks import ModelCheckpoint
from tensorflow.keras.callbacks import EarlyStopping
import os
import shutil
import random
from pathlib import Path
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
```

```python
#Data setup
train_dir = '/content/train_data/train'
test_dir = '/content/test_data/test'
```

```python
#Data setup
original_train_dir = Path('/content/train_data/train')
new_base_dir = Path('content/split')
new_train_dir = new_base_dir / "train"
new_val_dir = new_base_dir / "validation"
```

```python
# Parameters
val_split = 0.2  # 20% validation
SEED = 123
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Create new directories
for category in ["up", "down"]:
    os.makedirs(new_train_dir / category, exist_ok=True)
    os.makedirs(new_val_dir / category, exist_ok=True)

    # Get and shuffle files
    files = list((original_train_dir / category).glob("*.png"))
    random.shuffle(files)

    # Split
    split_idx = int(len(files) * (1 - val_split))
    train_files = files[:split_idx]
    val_files = files[split_idx:]

    # Copy files
    for f in train_files:
        shutil.copy(f, new_train_dir / category / f.name)
    for f in val_files:
        shutil.copy(f, new_val_dir / category / f.name)

print("✅ Split complete!")
print(f"Train set: {sum(len(list((new_train_dir / c).glob('*.png'))) for c in ['up', 'down'])} images")
print(f"Validation set: {sum(len(list((new_val_dir / c).glob('*.png'))) for c in ['up', 'down'])} images")
```

```python
#Rescale pixel values and augment training images
train_datagen = ImageDataGenerator(rescale=1./255, rotation_range=3, zoom_range=0.05, horizontal_flip=False,
                                   width_shift_range=0.1, height_shift_range=0.1, brightness_range=(0.9, 1.1))

val_datagen = ImageDataGenerator(rescale=1./255)

#Load images
train_generator = train_datagen.flow_from_directory(
    new_train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary')

val_generator = val_datagen.flow_from_directory(
    new_val_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary')
```

```python
#Model definition
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(224, 224, 3)),
    MaxPooling2D(2, 2),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')  # Binary classification
])
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
#model.compile(optimizer="rmsprop",
#    loss="sparse_categorical_crossentropy",
#    metrics=["accuracy"])
```

```python
#Save best model
checkpoint_cb = ModelCheckpoint('best_convnet_from_scratch.keras', save_best_only=True)
```

```python
#Epochs automasation
early_stopping = EarlyStopping(
    monitor='val_loss',      # or 'val_accuracy'
    patience=5,              # how many epochs to wait for improvement
    restore_best_weights=True)
```

```python
#Train
history = model.fit(train_generator,
                    epochs=10,
                    validation_data=val_generator,
                    callbacks=[checkpoint_cb, early_stopping])
```

```python
# Plot accuracy and loss
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], 'bo-', label='Training acc')
plt.plot(history.history['val_accuracy'],'ro-', label='Validation acc')
plt.title('Training and validation accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], 'bo-', label='Training loss')
plt.plot(history.history['val_loss'], 'ro-', label='Validation loss')
plt.title('Training and validation loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()

val_loss, val_acc = model.evaluate(val_generator)
print(f"Validation Loss: {val_loss:.4f}")
print(f"Validation Accuracy: {val_acc:.4f}")

#for epoch in range(len(history.history['loss'])):
    #print(f"Epoch {epoch+1}")
    #print(f"  Train   - Loss: {history.history['loss'][epoch]:.4f}, Accuracy: {history.history['accuracy'][epoch]:.4f}")
    #print(f"  Val     - Loss: {history.history['val_loss'][epoch]:.4f}, Accuracy: {history.history['val_accuracy'][epoch]:.4f}")
```

```python
from tensorflow.keras.applications import MobileNetV2, ResNet50, EfficientNetB0, VGG16
from tensorflow.keras import layers, models, Input
```

```python
checkpoint_cb_resnet = ModelCheckpoint('best_resnet50.keras', save_best_only=True)
checkpoint_cb_mobnet = ModelCheckpoint('best_mobilenetv2.keras', save_best_only=True)
checkpoint_cb_effnet = ModelCheckpoint('best_efficientnetb0.keras', save_best_only=True)
checkpoint_cb_vgg = ModelCheckpoint('best_vgg16.keras', save_best_only=True)
```

###ResNet50

```python
base_model_resnet = ResNet50(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet')

base_model_resnet.trainable = False
```

```python
inputs = Input(shape=(224, 224, 3))
x = base_model_resnet(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(1, activation='sigmoid')(x)

model_resnet = models.Model(inputs, outputs)
```

```python
model_resnet.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
base_model_resnet.trainable = True

# Fine-tune
fine_tune_at = 100
for layer in base_model_resnet.layers[:fine_tune_at]:
    layer.trainable = False

model_resnet.compile(optimizer=tf.keras.optimizers.Adam(1e-5),
              loss='binary_crossentropy',
              metrics=['accuracy'])

history_fine_resnet = model_resnet.fit(
    train_generator,
    epochs=10,
    validation_data=val_generator,
    callbacks=[checkpoint_cb_resnet, early_stopping])
```

##Evaluate all models

```python
#Evaluation Function
def evaluate_and_plot_confusion(model, val_generator, class_labels=['Down', 'Up'], model_name='Model'):
    # Reset the generator to start from the beginning
    val_generator.reset()

    # Predict probabilities for the validation set
    y_pred_probs = model.predict(val_generator)

    # Convert probabilities to binary predictions (0 or 1)
    y_pred_classes = (y_pred_probs > 0.5).astype("int32").flatten()

    # True labels from the generator
    y_true = val_generator.classes

    # Compute confusion matrix
    cm = confusion_matrix(y_true, y_pred_classes)

    # Plot confusion matrix heatmap
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_labels, yticklabels=class_labels)
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title(f'{model_name} - Confusion Matrix')
    plt.tight_layout()
    plt.show()

    # Print detailed classification report
    print(classification_report(y_true, y_pred_classes, target_names=class_labels))

# For ResNet50
evaluate_and_plot_confusion(model_resnet, val_generator, model_name="ResNet50")

# For MobileNetV2
evaluate_and_plot_confusion(model_mobnet, val_generator, model_name="MobileNetV2")

# For EfficientNetB0
evaluate_and_plot_confusion(model_effnet, val_generator, model_name="EfficientNetB0")

# For VGG16
evaluate_and_plot_confusion(model_vgg, val_generator, model_name="VGG16")

# For your from-scratch model
evaluate_and_plot_confusion(model, val_generator, model_name="From Scratch ConvNet")
```

##Loading test data and evaluation

```python
#Evaluation Function
def test_and_plot_confusion(model, generator, class_labels=['Down', 'Up'], model_name='Model'):
    # Reset the generator to start from the beginning
    generator.reset()

    # Predict probabilities for the validation set
    y_pred_probs = model.predict(generator)

    # Convert probabilities to binary predictions (0 or 1)
    y_pred_classes = (y_pred_probs > 0.5).astype("int32").flatten()

    # True labels from the generator
    y_true = generator.classes

    # Compute confusion matrix
    cm = confusion_matrix(y_true, y_pred_classes)

    # Plot confusion matrix heatmap
    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_labels, yticklabels=class_labels)
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title(f'{model_name} - Confusion Matrix')
    plt.tight_layout()
    plt.show()

    # Print detailed classification report
    print(classification_report(y_true, y_pred_classes, target_names=class_labels))
```

```python
#ROC-AUC Curve
from sklearn.metrics import roc_curve, auc

def plot_roc_auc(model, generator, model_name="Model"):
    y_true = generator.classes
    y_probs = model.predict(generator).ravel()

    fpr, tpr, _ = roc_curve(y_true, y_probs)
    roc_auc = auc(fpr, tpr)

    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, label=f'{model_name} (AUC = {roc_auc:.2f})', color='darkorange')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'ROC Curve - {model_name}')
    plt.legend()
    plt.tight_layout()
    plt.show()

# Example for all models
plot_roc_auc(model_resnet, test_generator, "ResNet50")
plot_roc_auc(model_mobnet, test_generator, "MobileNetV2")
plot_roc_auc(model_effnet, test_generator, "EfficientNetB0")
plot_roc_auc(model_vgg, test_generator, "VGG16")
plot_roc_auc(model_scratch, test_generator, "From Scratch")
```
